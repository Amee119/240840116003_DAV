# Practical: 4 | Perform Exploratory Data Analysis (EDA) using descriptive statistics and identify trends in datasets.

**Step 1: Import Required Libraries.**

```python
#import the pandas library.
#Pandas is used for reading, manipulating, and analyzing datasets.
import pandas as pd

#import the matplotlib library for creating graphs and charts.
import matplotlib.pyplot as plt
```

**Step 2: Create a Sample Dataset.**

```python
Data={
    "Month":["Januray","february","March","April","May","June","July","August","September","October","November","December"],
    "Units_Sold":[120,135,150,145,160,175,180,170,190,210,230,250],
    "Revenue":[120000,135000,150000,145000,160000,175000,180000,170000,190000,210000,230000,250000],
    "Profit":[20000,23000,25000,24000,27000,30000,31000,29000,33000,37000,41000,45000]
}
```

**Step 3: Create a DataFrame.**

```python
df = pd.DataFrame(Data)

# Display the complete DataFrame.
df
```

**Step 4: Inspect the Dataset.**

```python
#Task 1: Display the first five records.

df.head()
```

```python
#Task 2: Display the number of rows and columns.
df.shape
```

```python
#Task 3: Display all Columns.
df.columns
```

```python
#Task 4: Display information about the dataset.
df.info()
```

```python
#Task 5: Generate descriptive statistics for numerical columns.
df.describe()
```

**Step 5: Calculate Means.**

```python
#Calculate the avg number of units sold per month.
average_units=df["Units_Sold"].mean()

#Display the result.
print("Average units sold:",average_units)
```

**Step 6: Calculate Median.**

```python
#Calculate the middle value of the units_sold column.

median_units=df["Units_Sold"].median()

#Display the result
print("Median units sold:",median_units)
```

**Step 7: Calculate maximum and minimum.**

```python
#Find the highest number of units sold in a month.
maximum_units=df["Units_Sold"].max()

#Display the result.
print("Maximum units sold:",maximum_units)

print()
```

**Step 8: Calculate standard deviation.**

```python
#Calculate the standard deviation of units sold.
std_units = df["Units_Sold"].std()

 #Display the Result.
print("Standard deviation of units sold:",std_units)
```

**Step 9: Find the month with highest sales.**

```python
#Find the index of the row having the maximum revenue.

highest_sales_index=df["Revenue"].idxmax()

#Retrieve the month corresponding to the index.
highest_sales_month=df.loc[highest_sales_index,"Month"]

#Retrieve the maximum revenue.
highest_revenue=df.loc[highest_sales_index,"Revenue"]

#Display the result.
print("Month with highest sales:",highest_sales_month)
print("Highest revenue:",highest_revenue)
```

**Step 10 : Find the month with lowest sales**

```python
#Find the index of the row having the lowest revenue.

lowest_sales_index = df["Revenue"].idxmin()

#Retrieve the month corresponding to the index.
lowest_sales_month = df.loc[lowest_sales_index,"Month"]

#Retrieve the lowest revenue.
lowest_revenue = df.loc[lowest_sales_index,"Revenue"]

#Display the result.
print("Month with lowest sales:",lowest_sales_month)
print("Lowest revenue:",lowest_revenue)
```

**Step 11: Caluclate total revenue.**

```python
#Calculate the total revenue generated during the year.
total_revenue=df["Revenue"].sum()

#Display the result.
print("Total revenue generated:",total_revenue)
```

**Step 12: Calculate total profit.**

```python
#Calculate the total profit.
total_profit=df["Profit"].sum()

#Display the result.
print("Total profit:",total_profit)
```

**Step 13: Identify the sales trend.**

```python
from matplotlib.lines import lineStyles

#Create a line chart to visulize monthy revenue.
plt.figure(figsize=(10,5))

#plot month on the  x-axis and revenue on the y-axis.
plt.plot(df["Month"], df["Revenue"], marker="o", linestyle ="--")

#Add a title to the graph.
plt.title("Monthly Revenue")

# Add labels to the x-axis and y-axis.
plt.xlabel("Month")
plt.ylabel("Revenue")

# Rotate the month labels by 45 degrees for better readability.
plt.xticks(rotation=45)

#display the graph.
plt.show()
```

**Step 14: Analyze profit trend.**

```python
from matplotlib.lines import lineStyles
#create a line chart to visulize monthy revenue.
plt.figure(figsize=(10,5))

# plot month on the  x-axis and revenue on the y-axis
plt.plot(df["Month"], df["Profit"], marker="o", linestyle ="--")

#Add a title to the graph
plt.title("Monthly Profit")

# Add labels to the x-axis and y-axis.
plt.xlabel("Month")
plt.ylabel("Profit")

# Rotate the month labels by 45 degrees for better readability.
plt.xticks(rotation=45)

#display the graph
plt.show()
```

**Step 15: Compare revenue and profit.**

```python
#create a line chart to compare revenue and profit.
plt.figure(figsize=(10,5))

#plot revenue.
plt.plot(df["Month"],df["Revenue"],marker="o",linestyle="--",label="Revenue")

#plot profit.
plt.plot(df["Month"],df["Profit"],marker="o",linestyle="--",label="Profit")

#Add a title to the graph.
plt.title("Monthly Revenue and Profit")

# Add labels to the x-axis and y-axis.
plt.xlabel("Month")
plt.ylabel("Amount")

#display the legend to identify the two line.
plt.legend()

# Rotate the month labels by 45 degrees for better readability.
plt.xticks(rotation=45)

#display the graph.
plt.show()
```

**step 16: Identify growth in sales.**

```python
#calculate the percentage change in revenue.
#from one month to the next.
df["Revenue_Change"]=df["Revenue"].pct_change()*100

#display the update dataset.
df
```

**Step 17: Identify months with increasing revenue.**

```python
#Filter the dataset to display only those months.
#Where revenue growth is positive.
increasing_sales=df[df["Revenue_Change"]>0]

#Display the result
print("Months with increasing revenue:")
print(increasing_sales["Month"])
```
