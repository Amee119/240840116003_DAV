## **Practical-6 Create basic data visualizations (Line, Bar, Histogram, Scatter, Pie).**


Step 1 : Import require libraries numpy, matplotlib.

```python
# Import Pandas for data manipulation and analysis
import pandas as pd
# Import NumPy for numerical calculations
import numpy as np
# Import Matplotlib for creating graphs and visualizations
import matplotlib.pyplot as plt
# Import Seaborn for statistical data visualization
import seaborn as sns
```

Step 2: Create a dataset.

```python
# Create a Dataset
data = {
    "Month":["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"],
    "Product":["Laptop","Mobile","Tablet","Laptop","Mobile","Tablet","Laptop","Mobile","Tablet","Laptop","Mobile","Tablet"],
    "Region":["North","South","East","West","North","South","East","West","North","South","East","West"],
    "Sales":[120000,135000,150000,145000,160000,175000,180000,170000,190000,210000,230000,250000],
    "Units_Sold":[12,30,25,14,35,30,18,38,32,21,45,40]
}
# Convert the dictionary into a Pandas DataFrame
df =pd.DataFrame(data)
# Display the DataFrame
df
```

Task 2 Create a line Chart

2.1 How did the company sales change from jan to dec




```python
# Create a figure
plt.figure(figsize=(8, 4))
# Plot Month vs Sales
plt.plot(df['Month'], df['Sales'], marker='o', linestyle='-', color='r')
# Set chart title
plt.title('Company Sales Change from January to December')
# Set X-axis label
plt.xlabel('Month')
# Set Y-axis label
plt.ylabel('Total Sales')
# Show grid lines
plt.grid(True)
# Rotate month names for better readability
plt.xticks(rotation=45)
# Adjust layout
plt.tight_layout()
# Display the chart
plt.show()
```

Task 3 Create a Bar chart

3.1 Which product generates how much total sales(Groupby)

```python
# Group data by Product and calculate total Sales
product_sales = df.groupby('Product')['Sales'].sum().reset_index()
# Create a figure
plt.figure(figsize=(8, 4))
# Create a bar chart
sns.barplot(x='Product', y='Sales', data=product_sales, palette='viridis')
# Set chart title
plt.title('Total Sales by Product')
# Set X-axis label
plt.xlabel('Product')
# Set Y-axis label
plt.ylabel('Total Sales')
# Show grid lines on Y-axis
plt.grid(axis='y', linestyle='--', alpha=0.7)
# Adjust layout
plt.tight_layout()
# Display the chart
plt.show()
```

Task 4 Create a Histogram

4.1 How are the individual sales values distributed

```python
# Create a figure
plt.figure(figsize=(8, 4))
# Create a histogram for Sales data
sns.histplot(df['Sales'], bins=5, color='yellow')
# Set chart title
plt.title('Distribution of Individual Sales Values')
# Set X-axis label
plt.xlabel('Sales')
# Set Y-axis label
plt.ylabel('Frequency')
# Show grid lines on Y-axis
plt.grid(axis='y', alpha=0.7)
# Adjust layout
plt.tight_layout()
# Display the chart
plt.show()
```

Task 5 Create a Scatter plot

5.1 Is there are relationship between the number of unit sold and total sales

```python
# Create a figure
plt.figure(figsize=(8, 4))
# Create a scatter plot
plt.scatter(df['Units_Sold'], df['Sales'], color='Blue', alpha=0.7)
# Set chart title
plt.title('Relationship Between Units Sold and Total Sales')
# Set X-axis label
plt.xlabel('Units Sold')
# Set Y-axis label
plt.ylabel('Total Sales')
# Show grid lines
plt.grid(True)
# Adjust layout
plt.tight_layout()
# Display the chart
plt.show()
```

Task 6 Create a Pie Chart

6.1 What % of the total sales comes from each region

```python
# Group data by Region and calculate total Sales
region_sales = df.groupby('Region')['Sales'].sum()
# Create a figure
plt.figure(figsize=(4, 4))
# Create a pie chart
plt.pie(region_sales, labels=region_sales.index, autopct='%1.1f%%', startangle=140, colors=sns.color_palette('pastel'))
# Set chart title
plt.title('Percentage of Total Sales by Region')
# Make the pie chart circular
plt.axis('equal')
# Adjust layout
plt.tight_layout()
# Display the chart
plt.show()
```
