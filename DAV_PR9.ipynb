**Step 2: Import libraries**

```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go

from dash import Dash, dcc, html, Input, Output, State
from google.colab import files
from IPython.display import display

print("All libraries imported successfully!")
```

**Step 3: Create the McDonald's sample dataset**

```python
import numpy as np
import pandas as pd
np.random.seed(42)
n = 600
dates = pd.date_range(
    start="2025-01-01",
    end="2025-12-31",
    periods=n
)
cities = ["Ahmedabad", "Mumbai", "Delhi",
          "Pune", "Bengaluru"]
products = {
    "Burgers": ["McAloo Tikki", "McVeggie",
                "McChicken", "Maharaja Mac"],
    "Sides": ["French Fries", "Chicken Nuggets",
              "Veg Pizza McPuff"],
    "Beverages": ["Coke", "Cold Coffee", "Iced Tea"],
    "Desserts": ["McFlurry", "Soft Serve", "Brownie"]
}
prices = {
    "McAloo Tikki": 99,
    "McVeggie": 159,
    "McChicken": 179,
    "Maharaja Mac": 249,
    "French Fries": 89,
    "Chicken Nuggets": 149,
    "Veg Pizza McPuff": 59,
    "Coke": 65,
    "Cold Coffee": 119,
    "Iced Tea": 89,
    "McFlurry": 99,
    "Soft Serve": 49,
    "Brownie": 89
}
costs = {
    "McAloo Tikki": 55,
    "McVeggie": 92,
    "McChicken": 105,
    "Maharaja Mac": 150,
    "French Fries": 38,
    "Chicken Nuggets": 82,
    "Veg Pizza McPuff": 25,
    "Coke": 22,
    "Cold Coffee": 48,
    "Iced Tea": 31,
    "McFlurry": 43,
    "Soft Serve": 18,
    "Brownie": 37
}
rows = []
for i in range(n):
    category = np.random.choice(list(products.keys()))
    product = np.random.choice(products[category])
    quantity = np.random.randint(1, 6)
    revenue = prices[product] * quantity
    cost = costs[product] * quantity
    profit = revenue - cost
    rows.append({
        "Order_ID": f"ORD{i+1:04d}",
        "Date": dates[i],
        "City": np.random.choice(cities),
        "Product": product,
        "Category": category,
        "Quantity": quantity,
        "Revenue": revenue,
        "Cost": cost,
        "Profit": profit,
        "Payment": np.random.choice(
            ["UPI", "Cash", "Card", "Online"]
        ),
        "Rating": round(np.random.uniform(3, 5), 1)
    })
df = pd.DataFrame(rows)
display(df.head(10))
print("Dataset created successfully!")
print("Total Records:", len(df))
```

**Step 4: Download CSV dataset**

```python
df.to_csv("mcdonalds_sales.csv", index=False)

files.download("mcdonalds_sales.csv")
```

**Step 5: Data cleaning**

```python
# Check missing values
print("Missing Values:")
print(df.isnull().sum())
# Check duplicate records
print("Duplicate Records:", df.duplicated().sum())
# Remove duplicate records
df = df.drop_duplicates()
# Handle missing values safely by checking if columns exist
if "Rating" in df.columns:
    df["Rating"] = df["Rating"].fillna(
        df["Rating"].mean()
    )
if "City" in df.columns:
    df["City"] = df["City"].fillna("Unknown")
# Check cleaned data
print("\nCleaned Dataset:")
display(df.head())
print("Total Rows:", len(df))
print("Data Cleaning Completed!")
```

**Step 6: Basic data analysis**

```python
print("Total Revenue:", df["Revenue"].sum())
print("Total Profit:", df["Profit"].sum())
print("Total Orders:", df["Order_ID"].nunique())
print("Average Rating:", round(df["Rating"].mean(), 2))
print("\nSales by Category:")
print(
    df.groupby("Category")["Revenue"].sum()
)
print("\nSales by City:")
print(
    df.groupby("City")["Revenue"].sum()
)
```

**Step 7: Create six individual interactive charts**

**Chart 1 – Monthly Sales Trend**

```python
df["Month"] = df["Date"].dt.month_name()
monthly_sales = df.groupby(
    "Month"
)["Revenue"].sum().reset_index()
fig1 = px.line(
    monthly_sales,
    x="Month",
    y="Revenue",
    markers=True,
    title="Monthly Sales Trend",
    template="plotly_white"
)
fig1.show()
```

**Chart 2 – Sales by Food Category**

```python
category_sales = df.groupby(
    "Category"
)["Revenue"].sum().reset_index()
fig2 = px.pie(
    category_sales,
    names="Category",
    values="Revenue",
    hole=0.4,
    title="Sales by Food Category"
)
fig2.show()
```

**Chart 3 – Top 5 Selling Products**

```python
top_products = df.groupby(
    "Product"
)["Quantity"].sum().reset_index()
top_products = top_products.nlargest(
    5, "Quantity"
)
fig3 = px.bar(
    top_products,
    x="Product",
    y="Quantity",
    title="Top 5 Selling Products",
    color="Product",
    template="plotly_white"
)
fig3.show()
```

**Chart 4 – Profit by City**

```python
city_profit = df.groupby(
    "City"
)["Profit"].sum().reset_index()
fig4 = px.bar(
    city_profit,
    x="City",
    y="Profit",
    title="Profit by City",
    color="City",
    template="plotly_white"
)
fig4.show()
```

**Chart 5 – Payment Method Distribution**

```python
payment_data = df.groupby(
    "Payment"
)["Order_ID"].count().reset_index()
fig5 = px.pie(
    payment_data,
    names="Payment",
    values="Order_ID",
    hole=0.4,
    title="Payment Method Distribution"
)
fig5.show()
```

**Chart 6 – Customer Ratings**

```python
rating_data = df.groupby(
    "Rating"
)["Order_ID"].count().reset_index()
fig6 = px.bar(
    rating_data,
    x="Rating",
    y="Order_ID",
    title="Customer Rating Distribution",
    template="plotly_white"
)
fig6.show()
```

**Step 8: Create the complete dashboard**

```python
from dash import Dash, dcc, html, Input, Output, State
import plotly.express as px
import pandas as pd
# Initialize Dash App
app = Dash(__name__)
app.title = "McDonald's Business Dashboard"
# Dashboard Styling
app.layout = html.Div(
    style={
        "backgroundColor": "#f4f6f9",
        "padding": "20px",
        "fontFamily": "Arial"
    },
    children=[
        # Dashboard Heading
        html.H1(
            "McDonald's Sales Dashboard",
            style={
                "textAlign": "center",
                "color": "#c8102e"
            }
        ),
        html.P(
            "Business Performance & Sales Analysis",
            style={"textAlign": "center"}
        ),
        html.Hr(),
        # Filters
        html.Div([
            html.Div([
                html.Label("Select City"),
                dcc.Dropdown(
                    id="city_filter",
                    options=[
                        {"label": "All Cities", "value": "All"}
                    ] + [
                        {"label": c, "value": c}
                        for c in sorted(df["City"].unique())
                    ],
                    value="All",
                    clearable=False
                )
            ], style={
                "width": "48%",
                "display": "inline-block"
            }),
            html.Div([
                html.Label("Select Category"),
                dcc.Dropdown(
                    id="category_filter",
                    options=[
                        {"label": "All Categories", "value": "All"}
                    ] + [
                        {"label": c, "value": c}
                        for c in sorted(df["Category"].unique())
                    ],
                    value="All",
                    clearable=False
                )
            ], style={
                "width": "48%",
                "display": "inline-block",
                "marginLeft": "3%"
            })
        ], style={"marginBottom": "20px"}),
        # KPI Cards
        html.Div(
            id="kpi_cards",
            style={
                "display": "grid",
                "gridTemplateColumns": "repeat(4, 1fr)",
                "gap": "15px"
            }
        ),
        html.Br(),
        # First Row
        html.Div([
            dcc.Graph(id="chart1"),
            dcc.Graph(id="chart2")
        ], style={
            "display": "grid",
            "gridTemplateColumns": "1fr 1fr",
            "gap": "15px"
        }),
        # Second Row
        html.Div([
            dcc.Graph(id="chart3"),
            dcc.Graph(id="chart4")
        ], style={
            "display": "grid",
            "gridTemplateColumns": "1fr 1fr",
            "gap": "15px"
        }),
        # Third Row
        html.Div([
            dcc.Graph(id="chart5"),
            dcc.Graph(id="chart6")
        ], style={
            "display": "grid",
            "gridTemplateColumns": "1fr 1fr",
            "gap": "15px"
        }),
        html.Br(),
        # Download Button
        html.Div([
            html.Button(
                "Download CSV",
                id="download_btn",
                n_clicks=0,
                style={
                    "backgroundColor": "#c8102e",
                    "color": "white",
                    "padding": "12px 25px",
                    "border": "none",
                    "borderRadius": "8px",
                    "cursor": "pointer",
                    "fontSize": "16px"
                }
            ),
            dcc.Download(id="download_data")
        ], style={"textAlign": "center"}),
        html.Br(),
        html.P(
            "Created using Python, Dash, Plotly and Pandas",
            style={"textAlign": "center", "color": "gray"}
        )
    ]
)
# Callback to Update Dashboard
@app.callback(
    Output("kpi_cards", "children"),
    Output("chart1", "figure"),
    Output("chart2", "figure"),
    Output("chart3", "figure"),
    Output("chart4", "figure"),
    Output("chart5", "figure"),
    Output("chart6", "figure"),
    Input("city_filter", "value"),
    Input("category_filter", "value")
)
def update_dashboard(city, category):
    # Filter Dataset
    filtered = df.copy()
    if city != "All":
        filtered = filtered[
            filtered["City"] == city
        ]
    if category != "All":
        filtered = filtered[
            filtered["Category"] == category
        ]
    # Calculate KPIs
    revenue = filtered["Revenue"].sum()
    orders = filtered["Order_ID"].nunique()
    profit = filtered["Profit"].sum()
    rating = (
        filtered["Rating"].mean()
        if len(filtered) > 0 else 0
    )
    # KPI Card Function
    def create_card(title, value):
        return html.Div([
            html.H4(title),
            html.H2(value, style={
                "color": "#c8102e"
            })
        ], style={
            "backgroundColor": "white",
            "padding": "15px",
            "borderRadius": "10px",
            "textAlign": "center",
            "boxShadow": "0 2px 5px #ddd"
        })
    cards = [
        create_card(
            "Total Revenue",
            f"₹{revenue:,.0f}"
        ),
        create_card(
            "Total Orders",
            f"{orders:,}"
        ),
        create_card(
            "Total Profit",
            f"₹{profit:,.0f}"
        ),
        create_card(
            "Average Rating",
            f"{rating:.1f}/5"
        )
    ]
    # Chart 1: Monthly Sales
    monthly = filtered.copy()
    monthly["Month"] = (
        monthly["Date"].dt.strftime("%b")
    )
    monthly["Month_Number"] = (
        monthly["Date"].dt.month
    )
    monthly_sales = monthly.groupby(
        ["Month", "Month_Number"],
        as_index=False
    )["Revenue"].sum().sort_values("Month_Number")
    fig1 = px.line(
        monthly_sales,
        x="Month",
        y="Revenue",
        markers=True,
        title="Monthly Sales Trend",
        template="plotly_white"
    )
    # Chart 2: Category Sales
    cat_sales = filtered.groupby(
        "Category",
        as_index=False
    )["Revenue"].sum()
    fig2 = px.pie(
        cat_sales,
        names="Category",
        values="Revenue",
        hole=0.4,
        title="Sales by Food Category"
    )
    # Chart 3: Top Products
    top = filtered.groupby(
        "Product",
        as_index=False
    )["Quantity"].sum()
    top = top.nlargest(5, "Quantity")
    fig3 = px.bar(
        top,
        x="Product",
        y="Quantity",
        title="Top 5 Selling Products",
        color="Product"
    )
    # Chart 4: City Profit
    city_data = filtered.groupby(
        "City",
        as_index=False
    )["Profit"].sum()
    fig4 = px.bar(
        city_data,
        x="City",
        y="Profit",
        title="Profit by City",
        color="City"
    )
    # Chart 5: Payment Method
    payment = filtered.groupby(
        "Payment",
        as_index=False
    )["Order_ID"].count()
    fig5 = px.pie(
        payment,
        names="Payment",
        values="Order_ID",
        hole=0.4,
        title="Payment Method Distribution"
    )
    # Chart 6: Customer Ratings
    rating_data = filtered.groupby(
        "Rating",
        as_index=False
    )["Order_ID"].count()
    fig6 = px.bar(
        rating_data,
        x="Rating",
        y="Order_ID",
        title="Customer Rating Distribution"
    )
    return (
        cards,
        fig1,
        fig2,
        fig3,
        fig4,
        fig5,
        fig6
    )
# CSV Download Callback
@app.callback(
    Output("download_data", "data"),
    Input("download_btn", "n_clicks"),
    State("city_filter", "value"),
    State("category_filter", "value"),
    prevent_initial_call=True
)
def download_csv(n, city, category):
    filtered = df.copy()
    if city != "All":
        filtered = filtered[
            filtered["City"] == city
        ]
    if category != "All":
        filtered = filtered[
            filtered["Category"] == category
        ]
    return dcc.send_data_frame(
        filtered.to_csv,
        "McDonalds_Filtered_Sales.csv",
        index=False
    )
print("Dashboard created successfully!")
```
